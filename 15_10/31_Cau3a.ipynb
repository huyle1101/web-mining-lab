```python
from pathlib import Path
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

files = Path('Data_Preprocessing').glob('*.txt')
texts = [f.read_text(encoding='utf-8') for f in files]

opts = dict(tokenizer=str.split, token_pattern=None, lowercase=False) 
```

```python
vec = CountVectorizer(**opts)
X = vec.fit_transform(texts)

df = pd.DataFrame(X.toarray(), columns=vec.get_feature_names_out())
print(f"Kích thước ma trận: {X.shape}")
print(df.iloc[:5, :15])

print(f'Top từ xuất hiện nhiều nhất: {df.sum().sort_values(ascending=False).head(20)}')
```
