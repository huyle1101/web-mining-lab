```python
from pathlib import Path
from collections import Counter
import matplotlib.pyplot as plt
from wordcloud import WordCloud
```

```python
files = Path('Data_Preprocessing').glob('*.txt')
texts = [f.read_text(encoding='utf-8') for f in files]
tokens = [t for s in texts for t in s.split()]
```

```python
# thống kê tần suất
freq = Counter(tokens)
print(f"Số văn bản: {len(texts)}")
print(f"Tổng số từ: {len(tokens)}")
print(f"Top 20 từ: {freq.most_common(20)}")
```

```python
# biểu đồ cột top 20

words, counts = zip(*freq.most_common(20))
plt.figure(figsize=(11,5))
plt.bar(words, counts)
plt.xticks(rotation=60, ha='right')
plt.title("Top 20 từ xuất hiện nhiều nhất")
plt.tight_layout()
plt.show()
```

```python
# word cloud 

wc = WordCloud(width=900, height=500, background_color='white')
wc.generate_from_frequencies({w.replace('_', ' '): c for w, c in freq.items()})
plt.figure(figsize=(10,5))

plt.imshow(wc, interpolation='bilinear')
plt.axis('off')
plt.show()
```
