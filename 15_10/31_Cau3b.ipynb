```python
from pathlib import Path
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

files = Path('Data_Preprocessing').glob('*.txt')
texts = [f.read_text(encoding='utf-8') for f in files]

opts = dict(tokenizer=str.split, token_pattern=None, lowercase=False) 
```

```python
vec2 = CountVectorizer(ngram_range=(2,2), **opts)
X2 = vec2.fit_transform(texts)
df2 = pd.DataFrame(X2.toarray(), columns=vec2.get_feature_names_out())

print(f"Bigram kích thước: {X2.shape}")
print("Top các từ xuất hiện")
print(df2.sum().sort_values(ascending=False).head(15))
```
