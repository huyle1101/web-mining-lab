```python
import pandas as pd
```

```python
# one-hot encoding
def one_hot_encoding(sentence):
    words = sentence.split()
    
    df = pd.DataFrame({'words':words}) # 1 column
    
    one_hot = pd.get_dummies(df['words'])
    
    result = pd.concat([df,one_hot], axis=1)
    
    return result
```

```python
sentences = [
    "both good and distinctive",
    "good but not distinctive",
    "distinctive but not good, or",
    "neither good nor distinctive"
]

for sentence in sentences:
    sentence = sentence.replace(',','')
    print(one_hot_encoding(sentence))
    print('\n')
```

```python
# bow
from sklearn.feature_extraction.text import CountVectorizer

sentences = [
    "both good and distinctive",
    "good but not distinctive",
    "distinctive but not good, or",
    "neither good nor distinctive"
]

vectorize = CountVectorizer()
X = vectorize.fit_transform(sentences)

print(vectorize.get_feature_names_out())
print('\n')
print(X.toarray())
```

```python
# n-grams
import nltk

sentences = [
    "both good and distinctive",
    "good but not distinctive",
    "distinctive but not good, or",
    "neither good nor distinctive"
]

def to_ngrams(sentence, n=1):
    n_grams = nltk.ngrams(nltk.word_tokenize(sentence), n)
    return [' '.join(gram) for gram in n_grams]

for sentence in sentences:
    print(sentence)
    print(to_ngrams(sentence))
    print(to_ngrams(sentence, n=2))
    print('\n')
```

```python
# n-grams vectorization
ngrams_vectorizer = CountVectorizer(ngram_range=(1,2)) 

# the columns of the result matrix is both 1-gram and 2-gram ordered alphabetically
X = ngrams_vectorizer.fit_transform(sentences)
X.toarray()
```

```python
# n-grams and euclidean similarity
from scipy.spatial.distance import euclidean
from itertools import combinations

# 4 sentences will be unpacked into 4 variables
ngrams_vectorizer = CountVectorizer(ngram_range=(2,2))
n1, n2, n3, n4 = ngrams_vectorizer.fit_transform(sentences).toarray()

# print(euc)

data = {'n1': n1, 'n2': n2, 'n3': n3, 'n4': n4}
pairs = list(combinations(data.keys(), 2))

# for pair in pairs:
#     print(*pair)
#     print(euclidean(pair)) # * = unpack operator

for a,b in pairs:
    print(a,b)
    print(euclidean(data[a], data[b]))
    print('\n')
```
