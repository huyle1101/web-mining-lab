```python
# cách 1
l1 = []
n = 20
for i in range(1,n+1):
    page_url = f'{base_url}-p{i}'
    l1.append(page_url)

l1
```

```python
# cách 2 (1)

l2 = ['https://vnexpress.net/bong-da-p1']

count = 1

while count < 20:
    base_url = f'https://vnexpress.net/bong-da-p{count}'
    
    # print(base_url)
    r = requests.get(base_url)
    if r.status_code == 200:
        page_urls = re.findall(r'/bong-da-p\d+', r.text)
        tmp = set(page_urls)

        for url in tmp:
            s = 'https://vnexpress.net'+url
            if s not in l2:
                l2.append(s)
        count+=1

l2
```
