```python
import underthesea
import emot
import re
```

```python
text_to_process = '''HÔM NAY mình nhận đc hàng từ shop rồi nèee, áo ĐẸP LẮM nha mọi người ơi!!! 😍❤️. Nhưng mà giao hàng hơi trễ mất 3 ngày liền, làm mình chờ mún xỉu luôn :((. Mọi người có thể vào link https://shopee.vn/item-demo-12345 để xem chi tiết mẫu mã áo sơ mi này nhé. Giá gốc tận 250.000đ mà săn sale còn có 120k thôi, quá là rẻ lun í!. Chất vải dày dặn, ko bị xù lông....Tuy nhiên đường may chỗ cổ áo hơi bị lỗi một chút xíu, hi vọng shop sẽ cẩn thận hơn ở các đơn sau nhé :D 👍.'''
```

```python
def remove_url(text):
    res = re.sub(r'https?://\S+', '', text)
    return res

text = remove_url(text_to_process)
text
```

```python
# normalize
normalize = {
    'nèee':'nè',
    'mún':'muốn',
    'í': 'ấy',
    'ko':'không',
    'đc':'được'
}

for i, j in normalize.items():
    text = text.replace(i, j)
text
```

```python
emot_obj = emot.core.emot()

def convert_emotion(text):
    emoji = emot_obj.emoji(text)
    emoticons = emot_obj.emoticons(text)
    
    # replace emojis
    for i,j in zip(emoji['value'], emoji['mean']):
        text = text.replace(i,j.replace(':', ' '))
        
    # replace emoticons
    for i,j in zip(emoticons['value'], emoticons['mean']):
        text = text.replace(i,'_'.join(tmp for tmp in j.split()))
    
    return text

text = convert_emotion(text)
text
```

```python
text = text.lower()
text
```

```python
def remove_punctuation(text):
    res = re.sub(r'[^\w\s]*', '', text)
    return res

text = remove_punctuation(text)
text
```

```python
token_list = underthesea.word_tokenize(text)
token_list
```

```python
def remove_stop_words(list):
    with open('Data/vietnamese-stopwords.txt', 'r', encoding='utf-8') as f:
        stop_words = set(f.read().split('\n'))
    
    return [i for i in list if i not in stop_words]

result_list = remove_stop_words(token_list)
result_list 
```
