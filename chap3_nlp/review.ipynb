```python
import json
import re
from selenium import webdriver
from selenium.webdriver.common.by import By
import nltk
from nltk.corpus import stopwords

stop_words = set(stopwords.words('english'))
```

```python
driver = webdriver.Chrome()

url = 'https://en.wikipedia.org/wiki/Artificial_intelligence'
driver.get(url)
```

```python
text = driver.find_element(By.CSS_SELECTOR, 'div.mw-content-ltr.mw-parser-output').text
```

```python
text = text.lower()
```

```python
def remove_punctuations(text):
    s = re.sub(r'[^\w\s]*', '', text)
    return s
```

```python
text = remove_punctuations(text)
text
```

```python
def remove_digits(text):
    s = re.sub(r'\d+|\[\d+\]', '', text)
    return s
```

```python
text = remove_digits(text)
```

```python
def tokenize(text):
    return nltk.word_tokenize(text)
```

```python
tokenization = tokenize(text)
tokenization
```

```python
def remove_stop_words(tokinization):
    l = []
    for word in tokenization:
        if word.lower() not in stop_words:
            l.append(word)
    return l
```

```python
tokenization_no_stopwords = remove_stop_words(tokenization)
tokenization_no_stopwords
```

```python
with open('text.txt', 'w', encoding='utf-8') as f:
    f.write(text)
```
