```python
import json
import re
from selenium import webdriver
from selenium.webdriver.common.by import By
```

```python
driver = webdriver.Chrome()

url = 'https://en.wikipedia.org/wiki/Artificial_intelligence'
driver.get(url)
```

```python
text = driver.find_element(By.CSS_SELECTOR, 'div.mw-content-ltr.mw-parser-output').text
```

```python
with open('text.txt', 'w', encoding='utf-8') as f:
    f.write(text)
```

```python
def remove_html_tags(r_text):
    s = re.sub(r'<[^>]+>|', '', r_text)
    return s
```

```python
def remove_wiki_templates(r_text):
    s = re.sub(r'{{[^}]+}}', '', r_text)
    return s
```
