```python
import json
import re
from selenium import webdriver
from selenium.webdriver.common.by import By
import nltk
from nltk.corpus import stopwords

stop_words = set(stopwords.words('english'))
from nltk import FreqDist
from wordcloud import WordCloud
import matplotlib.pyplot as plt
```

```python
driver = webdriver.Chrome()

url = 'https://en.wikipedia.org/wiki/Artificial_intelligence'
driver.get(url)
```

```python
text = driver.find_element(By.CSS_SELECTOR, 'div.mw-content-ltr.mw-parser-output').text
```

```python
text = text.lower()
```

```python
def remove_punctuations(text):
    s = re.sub(r'[^\w\s]*', '', text)
    return s
```

```python
text = remove_punctuations(text)
text
```

```python
def remove_digits(text):
    s = re.sub(r'\d+|\[\d+\]', '', text)
    return s
```

```python
text = remove_digits(text)
```

```python
def tokenize(text):
    return nltk.word_tokenize(text)
```

```python
tokenization = tokenize(text)
tokenization
```

```python
def remove_stop_words(tokinization):
    l = []
    for word in tokenization:
        if word.lower() not in stop_words:
            l.append(word)
    return l
```

```python
tokenization_no_stopwords = remove_stop_words(tokenization)
tokenization_no_stopwords
```

```python
with open('Cau1_txl.txt', 'w', encoding='utf-8') as f:
    f.write('\n'.join(tokenization_no_stopwords))
```

```python
word_frequency = FreqDist(tokenization_no_stopwords)
word_frequency
```

```python
word_frequency.most_common(20)
```

```python
text_joined = ' '.join(tokenization_no_stopwords)

word_cloud = WordCloud(
    width=800,
    height=400,
    background_color='white'
).generate(text_joined)


plt.figure(figsize=(10, 5))
plt.imshow(word_cloud, interpolation="bilinear")
plt.axis("off")
plt.show()
```
