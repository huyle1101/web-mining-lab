```python
import requests
import json
import re
```

```python
with open(r'f:/web-mining-lab/headers.txt', 'r') as f:
    headers = f.read()
```

```python
url = 'https://en.wikipedia.org/wiki/Artificial_intelligence'
r = requests.get(url, headers=json.loads(headers))
r.status_code
```

```python
text = r.text.lower()
```

```python
def remove_html_tags(r_text):
    s = re.sub(r'<[^>]+>', '', r_text)
    return s
```

```python
text = remove_html_tags(text)
```
