```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.options import Options

options = Options()
options.page_load_strategy = 'eager'
```

```python
driver = webdriver.Chrome()

main_url = 'https://phongtro123.com/tinh-thanh/da-nang'

def get_href(driver, main_url):
    driver.get(main_url)
    href_els = driver.find_elements(By.CLASS_NAME, 'line-clamp-2')

    url_list = []

    for el in href_els:
        href = el.get_attribute('href')
        if href is not None:
            url_list.append(href)
    
    return url_list
```

```python
def scrape_information(driver, url):
    data = {}
    driver.get(url)

    title_el = driver.find_element(By.CSS_SELECTOR, '.border-bottom.pb-4.mb-4')
    infor = title_el.text.split('\n')
    data['id'] = infor[11]
    data['title'] = infor[0]
    data['price'] = infor[1]
    data['area'] = infor[2]
    data['updated_at'] = infor[3]
    data['city'] = infor[7]
    data['location'] = infor[9]
    data['posted_at'] = infor[-3]
    data['due_at'] = infor[-1]

    description_el = driver.find_elements(By.CSS_SELECTOR, '.border-bottom.pb-3.mb-4')
    s=''
    for el in description_el:
        s += el.text
    data['description'] = s
    return data

scrape_information(driver, 'https://phongtro123.com/cho-thue-phong-tro-full-noi-that-486-pr712191.html')
```
