```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.options import Options
import json
options = Options()
options.page_load_strategy = 'eager'
```

```python
def get_href(driver, main_url):
    driver.get(main_url)
    href_els = driver.find_elements(By.CLASS_NAME, 'line-clamp-2')

    url_list = []

    for el in href_els:
        href = el.get_attribute('href')
        if href is not None:
            url_list.append(href)
    
    return url_list
```

```python
def scrape_information(driver, url):
    data = {}
    driver.get(url)

    title_el = driver.find_element(By.CSS_SELECTOR, '.border-bottom.pb-4.mb-4')
    infor = title_el.text.split('\n')
    data['id'] = infor[11]
    data['url']
    data['title'] = infor[0]
    data['price'] = infor[1]
    data['area'] = infor[2]
    data['updated_at'] = infor[3]
    data['city'] = infor[7]
    data['location'] = infor[9]
    data['posted_at'] = infor[-3]
    data['due_at'] = infor[-1]

    description_el = driver.find_elements(By.CSS_SELECTOR, '.border-bottom.pb-3.mb-4')
    s=''
    for el in description_el:
        s += el.text
    data['description'] = s
    driver.back()
    return data
```

```python
def save_data(path, data):
    with open(path, 'a', encoding='utf-8') as f:
        json.dump(data, f, ensure_ascii=False)
        f.write('\n')
```

```python
driver = webdriver.Chrome()
main_url = 'https://phongtro123.com/tinh-thanh/da-nang?orderby=mac-dinh'
post_scraped = 0
page=2
while post_scraped < 20:
    url_list = get_href(driver, main_url)
    for url in url_list:
        print(f'Scraping {url}')
        data = scrape_information(driver, url)
        if data is not None:
            print(f'Scrape successful for {url}')
            save_data('../data/phong_tro.jsonl', data)
        else:
            print(f'Scrape failed for {url}')
        post_scraped+=1
        print(f'Post scraped: {post_scraped}')

    page+=1
    main_url = f'{main_url}&page={page}'
    print(f'Currently on {main_url}')
```
