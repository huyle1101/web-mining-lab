```python
import requests
import re
from bs4 import BeautifulSoup
```

```python
def get_text(html_text):
    s = re.sub(r'<[^>]*>', '', html_text)
    return s
```

```python
def process_url(url):
    processed_url = re.sub(r'\W+','_', url)
    return processed_url
```

```python
def save_data(path, data):
    with open(path, 'w', encoding='utf-8') as f:
        f.write(data)
```

```python
def main(url):
    processed_url = process_url(url)
    r = requests.get(url)
    
    if r.status_code == 200:
        print('Saving HTML source')
        html_text = BeautifulSoup(r.text, 'html.parser')
        save_data(f'f:/web-mining-lab/web_crawling/bao_thanh_nien_output/{processed_url}.html', html_text)
        
        print('Saving text')
```
