```python
import requests
import re
from bs4 import BeautifulSoup
import os
```

```python
def get_text(html_text):
    s = re.sub(r'<[^>]*>', '', html_text)
    return s
```

```python
def process_url(url):
    processed_url = re.sub(r'\W+','_', url)
    return processed_url
```

```python
def get_inner_urls(r_text):
    inner_urls = re.findall('<link>([^<]*)</link>', r_text) # need () group for re to return elements in that group
    
    return inner_urls
```

```python
def save_data(path, data):
    with open(path, 'w', encoding='utf-8') as f:
        f.write(data)
```

```python
def main(r_text, url):
    processed_url = process_url(url)
    
    print(f'Saving HTML source, {url}')
    html_text = BeautifulSoup(r.text, 'html.parser')
    save_data(f'f:/web-mining-lab/web_crawling/bao_thanh_nien_output/{processed_url}.html', str(html_text))
    
    print(f'Saving text, {url}')
    text = get_text(r_text)
    save_data(f'f:/web-mining-lab/web_crawling/bao_thanh_nien_output/{processed_url}.txt', text)
```

```python
url = 'https://thanhnien.vn/rss/kinh-te.rss'
r = requests.get(url)


if r.status_code == 200:
    main(r.text, url)
    inner_urls = get_inner_urls(r.text)
    for inner_url in inner_urls:
        r = requests.get('https://thanhnien.vn/rss/kinh-te.rss')
        if r.status_code == 200:
            html_text = BeautifulSoup(r.text, 'html.parser')
            main(r.text, inner_url)
        else:
            print(f'Request sent failed for inner url {inner_url}')


else:
    print(f'Request sent failed for url {url}')
```
