```python
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
import re

options = Options()
options.page_load_strategy = 'eager'
```

```python
def get_all_urls(driver):
    driver.find_elements(By.)
```

```python
def get_html_text(driver):
    return driver.page_source
```

```python
def process_url(url):
    processed_url = re.sub(r'[\W*]','_', url)
    return processed_url
```

```python
def save_data(path, data):
    with open(path, 'w', encoding='utf-8') as f:
        f.write(data)
```

```python
url = 'https://thanhnien.vn/rss/kinh-te.rss'

driver = webdriver.Chrome(options=options)
driver.get(url)
father_html_text = get_html_text(driver)
processed_url = process_url(url) 
save_data(f'f:/web-mining-lab/web_crawling/bao_thanh_nien_output/{processed_url}.html', father_html_text)
driver.quit()
```
